# 09 - Final Evaluation: GNSS Blackout Datasets

Phase 3 diagnostics for the GNSS-denied evaluation data. We inspect the blackout scenarios produced by `scripts/create_blackouts.py` from the Phase 2 calibrated trips: GNSS availability over time, the masked outage windows, the pre/blackout/post phases, and the offline vehicle reference trajectory that evaluation uses as ground truth (never given to the runtime model).

```python
import sys, os
from pathlib import Path
for p in ("..", "."):
    if (Path(p) / "src").is_dir():
        sys.path.insert(0, os.path.abspath(p)); break
```

```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from src.data.dataset_manager import DatasetManager

dm = DatasetManager("data/raw", "data/processed")
idx_path = Path("data/blackout/scenarios_index.json")
if not idx_path.is_file():
    raise SystemExit(
        "No blackout outputs yet. Run: python scripts/create_blackouts.py"
    )
index = pd.DataFrame(json.loads(idx_path.read_text(encoding="utf-8")))
print(f"{len(index)} blackout scenarios indexed")
index[['scenario_id','trip_id','duration_s','n_rows','total_masked_samples','has_reference']].head()
```

## 1. GNSS availability over time (runtime view)

The blackout cuts all GNSS fields (lat/lon/speed/accuracy/heading/satellites) inside the window; `gnss_available` drops to False there, IMU streams and timestamps are untouched.

```python
sid = "m__blk10s_00"
df = dm.load_blackout_scenario(sid)
runtime = dm.runtime_frame(df)
t = (df['timestamp'] - df['timestamp'].iloc[0]) / 60.0
bo = df['blackout_phase'] == 'blackout'

fig, ax = plt.subplots(2, 1, figsize=(11, 6), sharex=True)
ax[0].plot(t[~bo], runtime.loc[~bo, 'speed_kmh'], '.', ms=2, alpha=0.5, color='tab:blue')
ax[0].set_ylabel('GNSS speed (km/h)')
ax[0].set_title(f'{sid}: GNSS speed (masked during outage)')
ax[1].plot(t.values, df['gnss_available'].astype(int), drawstyle='steps-post', color='tab:green')
ax[1].set_ylabel('gnss_available')
ax[1].set_ylim(-0.1, 1.1)
ax[1].set_xlabel('time (min)')
for a in ax:
    a.axvspan(t[bo].min(), t[bo].max(), color='tab:red', alpha=0.18, label='blackout')
ax[0].legend(loc='upper right')
plt.tight_layout(); plt.show()
print('masked rows:', (~df['gnss_available']).sum(), '| phases:', df['blackout_phase'].value_counts().to_dict())
```

## 2. Offline reference trajectory vs observed GNSS

For trips with a synchronized vehicle reference (`vw16b`), the blackout dataset carries `reference_*` columns. Those are **evaluation-only ground truth** - the runtime frame drops them (see `DatasetManager.runtime_frame`).

```python
sid = "vw16b__blk10s_00"
df = dm.load_blackout_scenario(sid)
runtime = dm.runtime_frame(df)
ref = dm.reference_frame(df)
t = (df['timestamp'] - df['timestamp'].iloc[0]) / 60.0
bo = df['blackout_phase'] == 'blackout'

fig, ax = plt.subplots(2, 2, figsize=(12, 6))
ax[0, 0].plot(t[~bo], ref.loc[~bo, 'reference_latitude_deg'], color='tab:orange', label='reference (vehicle)')
ax[0, 0].plot(t[~bo], runtime.loc[~bo, 'latitude_deg'], '.', ms=2, color='tab:blue', label='GNSS (observed)')
ax[0, 0].set_title('latitude'); ax[0, 0].legend(fontsize=8)
ax[0, 1].plot(t[~bo], ref.loc[~bo, 'reference_speed_kmh'], color='tab:orange', label='reference')
ax[0, 1].plot(t[~bo], runtime.loc[~bo, 'speed_kmh'], '.', ms=2, color='tab:blue', label='GNSS')
ax[0, 1].set_title('speed (km/h)'); ax[0, 1].legend(fontsize=8)
ax[1, 0].plot(t.values, df['gnss_available'].astype(int), drawstyle='steps-post', color='tab:green')
ax[1, 0].set_title('gnss_available'); ax[1, 0].set_ylim(-0.1, 1.1)
ax[1, 1].axis('off')
for a in (ax[0, 0], ax[0, 1], ax[1, 0]):
    a.set_xlabel('time (min)')
    a.axvspan(t[bo].min(), t[bo].max(), color='tab:red', alpha=0.18, label='blackout')
ax[0, 0].legend(fontsize=8)
plt.tight_layout(); plt.show()
print('has_reference:', df['reference_available'].any(), '| source:', df['reference_source'].unique())
```

## 3. Masked samples per duration

Samples affected scale with the real timestamps (10 Hz nominal here: a 5 s outage cuts ~50 rows).

```python
rows = []
for sid in sorted(dm.list_blackout_scenarios()):
    if sid.startswith('m__'):
        m = dm.blackout_scenario_metadata(sid)
        rows.append({'duration_s': m['duration_s'], 'masked': m['realization']['total_masked_samples'], 'scenario': sid})
agg = pd.DataFrame(rows).sort_values('duration_s')
plt.figure(figsize=(8, 3.5))
plt.bar(agg['duration_s'].astype(str), agg['masked'], color='tab:red', alpha=0.8)
plt.xlabel('blackout duration (s)'); plt.ylabel('samples masked in trip m')
plt.title('Realized GNSS blackouts (deterministic seed 2024)')
plt.grid(axis='y', alpha=0.3); plt.tight_layout(); plt.show()
agg
```
